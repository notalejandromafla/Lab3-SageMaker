# Tarea 3.3 - Codificación de datos categóricos (Dataset Automóviles)

```python
import pandas as pd

# 1. Cargar el dataset de Automóviles desde UCI
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/autos/imports-85.data"
# Nombramos algunas columnas clave para el ejercicio (el archivo original no tiene encabezados)
columnas = ['symboling', 'normalized_losses', 'make', 'fuel_type', 'aspiration', 
            'num_of_doors', 'body_style', 'drive_wheels', 'engine_location']
# Leemos el archivo considerando "?" como valores nulos
df = pd.read_csv(url, names=columnas, usecols=range(9), na_values="?")

print("--- 1. Datos Originales (Muestra) ---")
display(df[['make', 'num_of_doors', 'body_style']].head())

# =====================================================================
# OBJETIVO 1: Codificar datos categóricos ORDINALES (Con orden lógico)
# 'num_of_doors' (two, four). Usamos un diccionario de mapeo.
# =====================================================================
mapeo_puertas = {'two': 2, 'four': 4}
df['num_of_doors_codificado'] = df['num_of_doors'].map(mapeo_puertas)

print("\n--- 2. Codificación Ordinal (num_of_doors a números) ---")
display(df[['num_of_doors', 'num_of_doors_codificado']].head())

# =====================================================================
# OBJETIVO 2: Codificar datos categóricos NO ORDINALES (Nominales)
# 'body_style' (sedan, hatchback). Usamos One-Hot Encoding (Variables Dummy)
# =====================================================================
df_onehot = pd.get_dummies(df, columns=['body_style'], prefix='estilo', dtype=int)

print("\n--- 3. Codificación No Ordinal (One-Hot Encoding para body_style) ---")
# Mostramos cómo la categoría se separó en columnas de 1s y 0s
columnas_estilo = [col for col in df_onehot.columns if 'estilo_' in col]
display(df_onehot[['make'] + columnas_estilo].head())
```
